## 1. Import Required Libraries
We first import the necessary libraries for data manipulation, visualization, and machine learning.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 2. Load and Explore Dataset
We load the dataset and explore its structure to understand its contents.

```python
df = pd.read_csv('/kaggle/input/spam-emails/spam.csv')
```

### Display First Few Rows
This helps us understand the format of the dataset.

### Check Dataset Information
We check for missing values, data types, and overall dataset structure.

## 3. Data Preprocessing
We will preprocess the text data before applying machine learning models. This includes:
- Text cleaning (removing special characters, converting to lowercase)
- Tokenization
- Removing stop words
- Applying TF-IDF transformation

```python
nlp = spacy.load('en_core_web_md')
```

```python
test_text = df.iloc[1400, 1]
test_text
```

```python
def preprocess(text):
    doc = nlp(text)
    filtered_data = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue
    
        filtered_data.append(token.lemma_)

    return " ".join(filtered_data)
```

```python
df['preprocessed message'] = df['Message'].apply(preprocess)
```

```python
encoder = LabelEncoder()

df['label'] = encoder.fit_transform(df['Category'])
```

```python
df.head()
```

```python
X = df['preprocessed message']
y = df['label']
```

```python
v = TfidfVectorizer()

X_vec = v.fit_transform(X)
```

```python
X_vec.toarray()
```

## 4. Splitting Data
We split the dataset into training and testing sets to evaluate model performance.

```python
X_train, X_test, y_train, y_test = train_test_split(X_vec, y, test_size=0.2, random_state=101)
```

```python
X_test.shape
```

## 5. Train Machine Learning Models
We will train a logistic regression model and a random forest classifier to classify emails as spam or ham.

```python
lr = LogisticRegression()

lr.fit(X_train, y_train)
```

```python
lr.score(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

## 6. Evaluate Model Performance
We evaluate our models using accuracy, confusion matrix, and classification report to understand their effectiveness.

```python
print(accuracy_score(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='0.0f', cmap='Blues')
plt.show()
```

```python
rf = RandomForestClassifier()

rf.fit(X_train, y_train)
```

```python
print(rf.score(X_train, y_train))
```

```python
y_pred = rf.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt="0.0f", cmap='Blues')
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(n_estimators=200)

xgb.fit(X_train, y_train)
```

```python
xgb.score(X_train, y_train)
```

```python
y_pred = xgb.predict(X_test)
```
